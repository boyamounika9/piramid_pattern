problems on class and object

```python
class circle :
    def __init__(self,radius):
        self.radius=radius
    def area(self):
        self.circlearea=3.14*(self.radius*self.radius)
        print(self.circlearea)
    def circum(self):
        self.circlecircum=2*3.14*self.radius
        print(self.circlecircum)

c=circle(12)
c.area()
c.circum()

```
