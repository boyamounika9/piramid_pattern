problems on class and object

```python
class circle :
    def __init__(self,radius):
        self.radius=radius
    def area(self):
        self.circlearea=3.14*(self.radius*self.radius)
        print(self.circlearea)
    def circum(self):
        self.circlecircum=2*3.14*self.radius
        print(self.circlecircum)

c=circle(12)
c.area()
c.circum()

```

```python
class student:
    def __init__(self,rollnum,studname,mark1,mark2,mark3):
        self.rollnum=rollnum
        self.studname=studname
        self.mark1=mark1
        self.mark2=mark2
        self.mark3=mark3
        
    def calculatetotal(self):
        self.totalmarks=self.mark1+self.mark2+self.mark3
        print(self.totalmarks)
    def displaystudentdetails(self):
        print(self.rollnum, self.studname, self.totalmarks)
s=student(13,"mounika",90,95,99)
s.calculatetotal()
s.displaystudentdetails()
```

```python
class account:
    def __init__(self,accountno,accounttype,accountbalance):
        self.givenamount=int(input("enter the amount:"))
        self.accountno=accountno
        self.accounttype=accounttype
        self.accountbalance=accountbalance
    def withdraw(self):
        self.balenceamount=self.accountbalance-self.givenamount
        print(self.balenceamount)
    def deposit(self):
        self.depo=self.accountbalance+self.givenamount
        print(self.depo)
    def dispaccountdetails(self):
        print(self.accountno,self.accounttype,self.accountbalance)
a=account(1,"zero",3000)
a.withdraw()
a.deposit()
a.dispaccountdetails()
```

```python
class rectangle:
    def __init__(self,length,breadth):
        self.length=length
        self.breadth=breadth
    def area(self):
        self.areaofrectangle=self.length*self.breadth
        print(self.areaofrectangle)
r=rectangle(30,20)
r.area()
```

problems on inheritance

single inheritance

```python
class shape:
    def __init__(self,radius):
        self.radius=radius
    def getperameter(self):
        print("parameter of shape")
    def getarea(self):
        print("area of shape")
class circle(shape):
    def getperameter(self):
        self.parameter=2*3.14*self.radius
        print(self.parameter)
    def getarea(self):
        self.area=3.14*(self.radius*self.radius)
        print(self.area)
c=circle(6)
c.getarea()
c.getperameter()
```

```python
class Shape:
    def area(self):
        print("This is shape area")


class Circle(Shape):
    def area(self):
        print("This is circle area")


class Rectangle(Shape):
    def area(self):
        print("This is rectangle area")



c = Circle()
r = Rectangle()

c.area()
r.area()
```

```python
class Shape:
    def area(self):
        print("Shape area")


class Circle(Shape):
    def area(self):
        print("Circle area")


class Rectangle(Shape):
    def area(self):
        print("Rectangle area")



def show_area(obj):
    obj.area()



c = Circle()
r = Rectangle()

show_area(c)
show_area(r)
```
